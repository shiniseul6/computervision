import os
import random
import shutil

def split_dataset():
    # 현재 위치(data.yaml이 있는 폴더)를 기준으로 경로 설정
    base_dir = '.'
    train_images_dir = os.path.join(base_dir, 'train', 'images')
    train_labels_dir = os.path.join(base_dir, 'train', 'labels')

    # 새롭게 만들어질 val 폴더 경로
    val_images_dir = os.path.join(base_dir, 'val', 'images')
    val_labels_dir = os.path.join(base_dir, 'val', 'labels')

    # val 폴더(images, labels) 생성
    os.makedirs(val_images_dir, exist_ok=True)
    os.makedirs(val_labels_dir, exist_ok=True)

    # 1. train 폴더에 있는 모든 이미지 목록 가져오기
    images = [f for f in os.listdir(train_images_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]
    
    # 2. 데이터를 무작위로 섞기
    random.seed(42)  # 항상 똑같이 섞이도록 시드 고정
    random.shuffle(images)

    # 3. 전체의 20%를 Validation(검증용)으로 계산
    val_size = int(len(images) * 0.2)
    val_images = images[:val_size]

    print(f"🚀 총 {len(images)}장의 데이터 중 {val_size}장을 val 폴더로 분리합니다...")

    # 4. 20%의 이미지와 라벨을 val 폴더로 이동(Move)
    for img_name in val_images:
        # 이미지 이동
        shutil.move(os.path.join(train_images_dir, img_name), os.path.join(val_images_dir, img_name))
        
        # 짝꿍인 라벨 파일(.txt) 이름 찾기
        label_name = os.path.splitext(img_name)[0] + '.txt'
        label_path = os.path.join(train_labels_dir, label_name)
        
        # 라벨 파일도 함께 이동
        if os.path.exists(label_path):
            shutil.move(label_path, os.path.join(val_labels_dir, label_name))

    print("✅ 데이터 스플릿(8:2 분할)이 완벽하게 끝났습니다! 폴더를 확인해 보세요.")

if __name__ == '__main__':
    split_dataset()
